# 🎬 Automated US History Shorts — ComfyUI (`LTX-Video 2B`) + HeyGen HyperFrames + Kaggle GPU Auto-Disconnect
- **GPU Quota Saver (`30 hrs/week`)**: Boots headless **ComfyUI** on Kaggle's free `Tesla T4 x2` / `P100` GPU, animates your 9:16 `MAI-Image-2.6-Flash` historical keyframes into **realistic moving cinema video clips**, renders the **HeyGen HyperFrames v13.0** composition, uploads the finished Short to YouTube, and **immediately disconnects/shuts down the GPU session** so zero idle minutes are ever wasted.
- **Setup**: In Kaggle, open **Settings -> Accelerator -> GPU T4 x2**, and add your secrets under **Add-ons -> Secrets** (or let GitHub Actions push this kernel automatically every 12 hours via `KAGGLE_USERNAME` + `KAGGLE_KEY`).

```python
# Cell 1: System Dependencies (FFmpeg, Node.js 22 for HyperFrames, and Repo Clone)
import os, subprocess, sys
from pathlib import Path

subprocess.run('apt-get update -qq && apt-get install -y -qq ffmpeg fonts-montserrat fonts-liberation fonts-dejavu-core libmagic1 libnss3 libgbm1', shell=True, check=False)
subprocess.run('curl -fsSL https://deb.nodesource.com/setup_22.x | bash - && apt-get install -y -qq nodejs', shell=True, check=False)

REPO_URL = os.getenv('GITHUB_REPO_URL', 'https://github.com/Rageshukla69/us-history-shorts-auto.git')
WORK_REPO = Path('/tmp/us-history-shorts-auto')
if not WORK_REPO.exists():
    subprocess.run(['git', 'clone', REPO_URL, str(WORK_REPO)], check=True)
else:
    subprocess.run(['git', '-C', str(WORK_REPO), 'pull', '--rebase'], check=False)

subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', str(WORK_REPO / 'requirements.txt')], check=True)
subprocess.run(['npm', 'install', '--no-audit', '--no-fund'], cwd=str(WORK_REPO), check=False)
print('✅ Repository & HyperFrames CLI ready at:', WORK_REPO)
```

```python
# Cell 2: Run 1 Complete ComfyUI + HyperFrames Short & Immediately Auto-Disconnect Kaggle GPU
import subprocess, sys
subprocess.run([sys.executable, '/tmp/us-history-shorts-auto/kaggle/run_kaggle_comfyui_short.py'], check=True)
```
